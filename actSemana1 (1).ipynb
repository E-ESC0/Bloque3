```python
import pandas as pd

colaboradores = pd.read_excel("DataBases_OTC.xlsx", sheet_name="Colaboradores")
df = pd.DataFrame(colaboradores)

# Limpiamos posibles espacios en blanco al inicio o al final de los nombres de las columnas
df.columns = df.columns.str.strip()

# Mostramos las columnas disponibles para verificar cuáles existen realmente
print("Columnas encontradas en el archivo:", list(df.columns))

# NOTA: Asegúrate de que los nombres coincidan exactamente con tu Excel.
# Si en tu Excel se llama de otra forma (ej. 'Sueldo Mensual' o 'Salario'), cambia los textos aquí:
col_salario = "Salario Diario" if "Salario Diario" in df.columns else "SalarioMensual"
col_sueldo = "Sueldo IMSS" if "Sueldo IMSS" in df.columns else "SueldoIMSS"
col_compensacion = "Compensacion" if "Compensacion" in df.columns else "Compensación"

def crearcolumnas():
  # Corregido: Calculamos el SalarioDiario usando la columna real existente
  # Si no existe 'SalarioMensual', busca la que corresponda en tu Excel
  if col_salario in df.columns:
    df["SalarioDiario"] = df[col_salario] / 160
  else:
    # Si no existe, podemos estimarlo provisionalmente o usar un valor existente
    df["SalarioDiario"] = df[col_sueldo] / 30 if col_sueldo in df.columns else 0

  if col_sueldo in df.columns and col_compensacion in df.columns:
    df["CostoInterno"] = df[col_sueldo] + df[col_compensacion]
  else:
    df["CostoInterno"] = 0
  return

def Limpieza():
  df.drop(df.columns[0], axis=1, inplace=True)
  return

def estadisticasreferencia():
  SalarioPromedio = df["SalarioDiario"].mean()
  SueldoPromedio = df[col_sueldo].mean() if col_sueldo in df.columns else 0
  CompensacionPromedio = df[col_compensacion].mean() if col_compensacion in df.columns else 0
  CostoInternoPromedio = df["CostoInterno"].mean()

  print("\n--- Estadísticas de Referencia ---")
  print(f"Salario Diario Promedio: {SalarioPromedio:.2f}")
  print(f"Sueldo Promedio: {SueldoPromedio:.2f}")
  print(f"Compensación Promedio: {CompensacionPromedio:.2f}")
  print(f"Costo Interno Promedio: {CostoInternoPromedio:.2f}\n")
  return

def actualiza():
  # Corregido: se guarda usando df.to_excel en lugar de pd.to_excel
  df.to_excel("archivocambiado.xlsx", index=False)
  print("¡Archivo guardado con éxito como 'archivocambiado.xlsx'!")
  return

crearcolumnas()
Limpieza()
estadisticasreferencia()
actualiza()
```
